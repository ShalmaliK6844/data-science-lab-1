```python
# Experiment 1: Data Collection
# Weather Data, Air Quality Data and Bookstore Web Scraping

print("Experiment 1: Data Collection")
```

```python
import requests
import pandas as pd
import time
from bs4 import BeautifulSoup
from pathlib import Path

print("All required libraries imported successfully.")
```

```python
DATA = Path("data")
DATA.mkdir(exist_ok=True)

print("Data folder:", DATA.resolve())
```

```python
# Five Bengaluru locations

locations = {
    "Bengaluru Central": (12.9716, 77.5946),
    "Whitefield": (12.9698, 77.7499),
    "Electronic City": (12.8452, 77.6602),
    "Yelahanka": (13.1007, 77.5963),
    "Jayanagar": (12.9250, 77.5938)
}

print("Number of locations:", len(locations))
print("\nLocations:")

for location, (latitude, longitude) in locations.items():
    print(f"{location}: ({latitude}, {longitude})")
```

```python
# Weather Data Collection
# Collect hourly temperature and humidity for 5 Bengaluru locations
# for the past 21 days

WEATHER_URL = "https://api.open-meteo.com/v1/forecast"

weather_data = []

for location, (latitude, longitude) in locations.items():

    print("Collecting weather data for:", location)

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": "temperature_2m,relative_humidity_2m",
        "timezone": "Asia/Kolkata",
        "past_days": 21
    }

    response = requests.get(
        WEATHER_URL,
        params=params,
        timeout=20
    )

    if response.status_code == 200:

        result = response.json()

        weather_df = pd.DataFrame(result["hourly"])

        weather_df["time"] = pd.to_datetime(weather_df["time"])

        weather_df["location"] = location
        weather_df["latitude"] = latitude
        weather_df["longitude"] = longitude

        weather_data.append(weather_df)

    else:
        print("Error for", location, response.status_code)

    time.sleep(1)


# Combine all 5 locations
bengaluru_weather = pd.concat(
    weather_data,
    ignore_index=True
)


# Rename columns
bengaluru_weather = bengaluru_weather.rename(columns={
    "temperature_2m": "temperature",
    "relative_humidity_2m": "humidity"
})


# Arrange columns
bengaluru_weather = bengaluru_weather[
    [
        "location",
        "latitude",
        "longitude",
        "time",
        "temperature",
        "humidity"
    ]
]


# Save the weather data
bengaluru_weather.to_csv(
    DATA / "bengaluru_weather.csv",
    index=False
)


print("\nWeather data collected successfully!")
print("Number of records:", len(bengaluru_weather))
print("Number of columns:", len(bengaluru_weather.columns))

print("\nFirst 5 records:")
print(bengaluru_weather.head())
```

```python
# Air Quality Data Collection
# Collect hourly PM10, PM2.5 and CO for 5 Bengaluru locations
# for the past 30 days

AIR_QUALITY_URL = "https://air-quality-api.open-meteo.com/v1/air-quality"

air_quality_data = []

for location, (latitude, longitude) in locations.items():

    print("Collecting air quality data for:", location)

    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": "pm10,pm2_5,carbon_monoxide",
        "timezone": "Asia/Kolkata",
        "past_days": 30
    }

    response = requests.get(
        AIR_QUALITY_URL,
        params=params,
        timeout=20
    )

    if response.status_code == 200:

        result = response.json()

        air_df = pd.DataFrame(result["hourly"])

        air_df["time"] = pd.to_datetime(air_df["time"])

        air_df["location"] = location
        air_df["latitude"] = latitude
        air_df["longitude"] = longitude

        air_quality_data.append(air_df)

    else:
        print("Error for", location, response.status_code)

    time.sleep(1)


# Combine all 5 locations
bengaluru_air_quality = pd.concat(
    air_quality_data,
    ignore_index=True
)


# Rename columns
bengaluru_air_quality = bengaluru_air_quality.rename(columns={
    "pm2_5": "pm2_5",
    "carbon_monoxide": "co"
})


# Arrange columns
bengaluru_air_quality = bengaluru_air_quality[
    [
        "location",
        "latitude",
        "longitude",
        "time",
        "pm10",
        "pm2_5",
        "co"
    ]
]


# Save the air quality data
bengaluru_air_quality.to_csv(
    DATA / "bengaluru_air_quality.csv",
    index=False
)


print("\nAir quality data collected successfully!")
print("Number of records:", len(bengaluru_air_quality))
print("Number of columns:", len(bengaluru_air_quality.columns))

print("\nFirst 5 records:")
print(bengaluru_air_quality.head())
```

```python
# Save collected weather and air quality data

weather_file = DATA / "bengaluru_weather.csv"
air_quality_file = DATA / "bengaluru_air_quality.csv"

bengaluru_weather.to_csv(weather_file, index=False)
bengaluru_air_quality.to_csv(air_quality_file, index=False)

print("Weather data saved to:", weather_file)
print("Air quality data saved to:", air_quality_file)
```

```python
# Bookstore web scraping

base_url = "https://books.toscrape.com/catalogue/page-1.html"

print("Bookstore URL:", base_url)
```

```python
books_data = []

print("Book data list initialized successfully.")
```

```python
# Bookstore web scraping

current_url = "https://books.toscrape.com/catalogue/page-1.html"

while current_url:
    print("Scraping:", current_url)

    response = requests.get(current_url)

    if response.status_code != 200:
        print("Failed to fetch page:", response.status_code)
        break

    soup = BeautifulSoup(response.text, "html.parser")

    # Extract book information
    books = soup.select("article.product_pod")

    for book in books:
        title = book.h3.a["title"]
        price = book.select_one(".price_color").text.strip()

        rating_class = book.select_one(".star-rating")["class"]
        rating = rating_class[1]

        books_data.append({
            "title": title,
            "price": price,
            "rating": rating
        })

    # Find next page
    next_button = soup.select_one("li.next a")

    if next_button:
        next_page = next_button["href"]

        if next_page.startswith("../"):
            current_url = "https://books.toscrape.com/catalogue/" + next_page.replace("../", "")
        else:
            current_url = "https://books.toscrape.com/catalogue/" + next_page
    else:
        current_url = None

print("\nBook scraping completed successfully!")
print("Number of books collected:", len(books_data))
```

```python
# Save scraped bookstore data to CSV

books_df = pd.DataFrame(books_data)

books_df.to_csv(DATA / "books_data.csv", index=False)

print("Book data saved successfully!")
print("File path:", DATA / "books_data.csv")
print("Number of records:", len(books_df))
print("Number of columns:", len(books_df.columns))

print("\nFirst 5 records:")
print(books_df.head())
```
